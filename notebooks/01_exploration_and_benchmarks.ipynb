# VisionX: Model Exploration, Quantization & Pipeline Benchmark
This notebook demonstrates loading YOLO models, running ByteTrack multi-object tracking, computing spatial analytics, and measuring FP32 vs FP16 vs INT8 inference latency.

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
from src.detection.synthetic_detector import SyntheticDetector
from src.tracking.tracker import ByteTracker
from src.analytics.zone_analytics import ZoneAnalytics
from src.detection.benchmarks import run_model_benchmarks
```

```python
# Initialize detector and tracker
detector = SyntheticDetector(conf_threshold=0.35)
tracker = ByteTracker(track_thresh=0.5, frame_rate=25)

# Create sample frame
frame = np.zeros((720, 1280, 3), dtype=np.uint8)
detections = detector.predict(frame)
tracks = tracker.update(detections)

print(f"Detected {len(detections)} objects, Active tracks: {len(tracks)}")
```

```python
# Run and display optimization benchmark matrix
benchmarks = run_model_benchmarks()
for b in benchmarks:
    print(f"{b.model_name:<40} | {b.precision:<6} | Latency: {b.p95_latency_ms:.1f}ms | {b.fps:.1f} FPS")
```
